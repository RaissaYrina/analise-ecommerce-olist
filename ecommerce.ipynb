### Tratando dados ausentes

```python
#convertendo tipo de data str para date

colunas_data =['order_purchase_timestamp',
                'order_approved_at',
                'order_delivered_carrier_date',
                'order_delivered_customer_date',
                'order_estimated_delivery_date'
]

for coluna in colunas_data:
    df_orders[coluna] = pd.to_datetime(
                            df_orders[coluna],
                            format = "%Y-%m-%d %H:%M:%S",
                            errors ="coerce",
    )
```

```python
df_orders[colunas_data].dtypes
```

```python
df_orders[colunas_data].isna().sum()
```

```python
sem_entrega = df_orders[df_orders["order_delivered_customer_date"].isna()]
sem_entrega["order_status"].value_counts()
```

```python
entregues = df_orders[
    (df_orders["order_status"] == "delivered")
    & (df_orders["order_delivered_customer_date"].notna())
].copy()

entregues["dias_entrega"] = (
    entregues["order_delivered_customer_date"]
    - entregues["order_purchase_timestamp"]
).dt.days

entregues["atrasado"] = (
    entregues["order_delivered_customer_date"]
    > entregues["order_estimated_delivery_date"]
)

entregues[["dias_entrega", "atrasado"]].describe(include="all")
```

```python
entregues.columns.tolist()
```

```python
entregues["atrasado"].mean() * 100
```

### Performance por estado

```python
entregues_estado = entregues.merge(
    df_customers[["customer_id", "customer_state"]],
    on="customer_id",
    how="left",
    validate="one_to_one",
)
```

```python
print(len(entregues), len(entregues_estado))
entregues_estado["customer_state"].isna().sum()
```

```python
por_estado = (
    entregues_estado
    .groupby("customer_state")
    .agg(
        pedidos=("order_id", "count"),
        dias_medio=("dias_entrega", "mean"),
        dias_mediana=("dias_entrega", "median"),
        pct_atraso=("atrasado", "mean"),
    )
    .assign(pct_atraso=lambda df: df["pct_atraso"] * 100)
    .round(1)
    .sort_values("dias_medio", ascending=False)
)

por_estado
```
