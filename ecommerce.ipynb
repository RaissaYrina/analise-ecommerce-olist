### Tratando dados ausentes

```python
#convertendo tipo de data str para date

colunas_data =['order_purchase_timestamp',
                'order_approved_at',
                'order_delivered_carrier_date',
                'order_delivered_customer_date',
                'order_estimated_delivery_date'
]

for coluna in colunas_data:
    df_orders[coluna] = pd.to_datetime(
                            df_orders[coluna],
                            format = "%Y-%m-%d %H:%M:%S",
                            errors ="coerce",
    )
```

```python
entregues = df_orders[
    (df_orders["order_status"] == "delivered")
    & (df_orders["order_delivered_customer_date"].notna())
].copy()

entregues["dias_entrega"] = (
    entregues["order_delivered_customer_date"]
    - entregues["order_purchase_timestamp"]
).dt.days

entregues["atrasado"] = (
    entregues["order_delivered_customer_date"]
    > entregues["order_estimated_delivery_date"]
)

entregues[["dias_entrega", "atrasado"]].describe(include="all")
```
